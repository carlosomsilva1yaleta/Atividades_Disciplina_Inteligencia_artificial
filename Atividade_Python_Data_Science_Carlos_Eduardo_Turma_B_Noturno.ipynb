Exercício 19 - Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).

```python
import numpy as np

array1 = np.random.randint(1, 101, 20)
array2 = np.random.randint(1, 101, 20)

produto_escalar = np.dot(array1, array2)

norma1 = np.sqrt(np.sum(array1 ** 2))
norma2 = np.sqrt(np.sum(array2 ** 2))

coseno = produto_escalar / (norma1 * norma2)

angulo = np.degrees(np.arccos(coseno))

print("Array 1:")
print(array1)

print("Array 2:")
print(array2)

print("Ângulo:", angulo, "graus")
```

Exercício 1 - Calcule a integral da função f(x) = x2 sin(x) no intervalo de 0 a π.

```python
import numpy as np

from scipy.integrate import quad

def f(x):
    return x**2 * np.sin(x)

resultado, erro = quad(f, 0, np.pi)

print("Integral:", resultado)
```

Exercício 2 - Resolva a equação diferencial dy dx = x − y, com y(0) = 1, no intervalo de 0 a 10.

```python
from scipy.integrate import solve_ivp
import numpy as np

def equacao(t, y):
    return t - y

solucao = solve_ivp(
    equacao,
    [0, 10],
    [1],
    t_eval=np.linspace(0, 10, 100)
)

print("Tempo:")
print(solucao.t)

print("Solução:")
print(solucao.y[0])
```

Exercício 3 - Encontre as raízes da função f(x) = x4 − 3x3 + 2 utilizando SciPy.

```python
from scipy.optimize import root_scalar

def f(x):
    return x**4 - 3*x**3 + 2

raiz1 = root_scalar(f, bracket=[0, 1])
raiz2 = root_scalar(f, bracket=[2, 3])

print("Raiz 1:", raiz1.root)
print("Raiz 2:", raiz2.root)
```

Exercício 4 - Use a função de otimização de SciPy para maximizar f(x) = −x^2+4x+1.

```python
from scipy.optimize import minimize

def funcao(x):
    return -( -x**2 + 4*x + 1 )

resultado = minimize(funcao, 0)

x = resultado.x[0]
maior_valor = -resultado.fun

print("Valor de x:", x)
print("Máximo:", maior_valor)
```

Exercício 5 - Calcule a transformada de Fourier de um sinal composto de si (2πt) e cos(4πt).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fftfreq

t = np.linspace(0, 1, 500)

sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)

transformada = fft(sinal)
frequencias = fftfreq(len(t), t[1] - t[0])

plt.plot(frequencias, np.abs(transformada))
plt.xlabel("Frequência")
plt.ylabel("Amplitude")
plt.title("Transformada de Fourier")
plt.grid()
plt.show()
```

Exercício 7 - Use a função de interpolação cúbica de SciPy para interpolar os dados (0, 0), (1,2),(2,3),(3,5),(4,4).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d

x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 2, 3, 5, 4])

funcao = interp1d(x, y, kind="cubic")

x_novos = np.linspace(0, 4, 100)
y_novos = funcao(x_novos)

plt.scatter(x, y, label="Pontos originais")
plt.plot(x_novos, y_novos, label="Interpolação cúbica")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Interpolação Cúbica")
plt.legend()
plt.grid()
plt.show()
```

Exercício 9 - Ajuste uma curva quadrática aos pontos (1,2),(2,4),(3,5),(4,7),(5,8).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])

def parabola(x, a, b, c):
    return a * x**2 + b * x + c

parametros, _ = curve_fit(parabola, x, y)

x_ajuste = np.linspace(1, 5, 100)
y_ajuste = parabola(x_ajuste, *parametros)

plt.scatter(x, y, label="Dados originais")
plt.plot(x_ajuste, y_ajuste, label="Curva ajustada")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Ajuste de Curva Quadrática")
plt.legend()
plt.grid()
plt.show()

print("Parâmetros [a, b, c]:")
print(parametros)
```

Exercício 10 - Calcule a integral dupla da função f(x,y) = xy + x2 no intervalo [0,2] para x e [0,1] para y.

```python
from scipy.integrate import dblquad

def f(y, x):
    return x * y + x**2

resultado, erro = dblquad(f, 0, 2, lambda x: 0, lambda x: 1)

print("Resultado da integral dupla:", resultado)
print("Erro estimado:", erro)
```

Exercício 16 - Resolva a equação diferencial de segunda ordem d2y/(dx2 +y) = 0, comcondições iniciais y(0) = 1 e y′(0) = 0.

```python
from scipy.integrate import solve_ivp
import numpy as np

def equacao(x, estado):
    y = estado[0]
    derivada = estado[1]

    return [derivada, -y]

solucao = solve_ivp(
    equacao,
    [0, 10],
    [1, 0],
    t_eval=np.linspace(0, 10, 100)
)

print("Valores de x:")
print(solucao.t)

print("Valores de y:")
print(solucao.y[0])
```

Exercício 6 - Plote dois gráficos de linha em subplots diferentes: y1 = ex e y2 = log(x).

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(1, 10, 100)

y1 = np.exp(x)
y2 = np.log(x)

plt.subplot(1, 2, 1)
plt.plot(x, y1)
plt.title("y = eˣ")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()

plt.subplot(1, 2, 2)
plt.plot(x, y2)
plt.title("y = log(x)")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()

plt.tight_layout()
plt.show()
```

Exercício 8 - Plote um gráfico de linha com intervalos de confiança (sombramento) para a função y = sin(x).

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

erro = 0.2

plt.plot(x, y, label="y = sen(x)")
plt.fill_between(x, y - erro, y + erro, alpha=0.3)

plt.title("Função y = sen(x) com intervalo de confiança")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()

plt.show()
```

Exercício 10 - Plote três gráficos de linha em um único gráfico, cada um com uma cor e estilo de linha diferentes.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)

plt.plot(x, y1, color='blue', linestyle='-', label='seno')
plt.plot(x, y2, color='red', linestyle='--', label='cosseno')
plt.plot(x, y3, color='green', linestyle=':', label='tangente')

plt.ylim(-2, 2)
plt.title('Três funções trigonométricas')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```

Exercício 13 - Gere um gráfico de dispersão 3D para 100 pontos aleatórios utilizando Matplotlib.

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

x = np.random.rand(100)
y = np.random.rand(100)
z = np.random.rand(100)

ax.scatter(x, y, z, c=z, cmap='viridis', marker='o')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Dispersão 3D')
plt.show()
```

Exercício 16 - Crie um gráfico de linha com títulos para cada eixo e adicione um rótulo de texto no ponto x = 5.

Exercício 17 - Plote um gráfico de linha e salve-o como um arquivo PDF.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.exp(-x/2) * np.sin(2*np.pi*x)

plt.plot(x, y)
plt.title('Gráfico de sinal')
plt.xlabel('tempo')
plt.ylabel('amplitude')
plt.grid(True)
plt.savefig('grafico.pdf', format='pdf')
plt.show()
```

Exercício 18 - Crie um gráfico de barras com valores positivos e negativos e use diferentes cores para cada tipo de valor.

```python
import matplotlib.pyplot as plt
import numpy as np

categorias = ['A', 'B', 'C', 'D', 'E']
valores = [10, -5, 8, -3, 6]

cores = ['green' if v >= 0 else 'red' for v in valores]

plt.bar(categorias, valores, color=cores)
plt.title('Barras com cores por sinal')
plt.xlabel('Categorias')
plt.ylabel('Valores')
plt.axhline(0, color='black', linewidth=0.8)
plt.show()
```
